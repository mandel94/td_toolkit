## 6. Content Length Categories

```python
# Categorize articles by length
def categorize_length(word_count):
    if word_count < 500:
        return 'Short'
    elif word_count < 1000:
        return 'Medium'
    elif word_count < 2000:
        return 'Long'
    else:
        return 'Very Long'

df['length_category'] = df['word_count'].apply(categorize_length)

# Category analysis
category_stats = df.groupby('length_category').agg({
    'word_count': ['count', 'mean'],
    'pageviews': 'mean',
    'engagement_rate': 'mean',
    'editorial_score': 'mean'
}).round(2)

category_stats.columns = ['Article Count', 'Avg Word Count', 'Avg Pageviews', 
                          'Avg Engagement Rate', 'Avg Editorial Score']

print("CONTENT LENGTH CATEGORIES")
print("="*80)
print(category_stats)
print("="*80)
```

```python
# Visualize category performance
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Article count by category
category_counts = df['length_category'].value_counts()
axes[0].bar(category_counts.index, category_counts.values, color=['#FF6B6B', '#4ECDC4', '#45B7D1', '#96CEB4'])
axes[0].set_xlabel('Length Category')
axes[0].set_ylabel('Number of Articles')
axes[0].set_title('Article Count by Length Category')
axes[0].grid(True, alpha=0.3, axis='y')

# Avg Editorial Score by category
category_score = df.groupby('length_category')['editorial_score'].mean().sort_values(ascending=False)
axes[1].bar(category_score.index, category_score.values, color=['#FF6B6B', '#4ECDC4', '#45B7D1', '#96CEB4'])
axes[1].set_xlabel('Length Category')
axes[1].set_ylabel('Avg Editorial Score')
axes[1].set_title('Average Editorial Score by Length Category')
axes[1].grid(True, alpha=0.3, axis='y')

# Avg Pageviews by category
category_views = df.groupby('length_category')['pageviews'].mean().sort_values(ascending=False)
axes[2].bar(category_views.index, category_views.values, color=['#FF6B6B', '#4ECDC4', '#45B7D1', '#96CEB4'])
axes[2].set_xlabel('Length Category')
axes[2].set_ylabel('Avg Pageviews')
axes[2].set_title('Average Pageviews by Length Category')
axes[2].grid(True, alpha=0.3, axis='y')

plt.tight_layout()
plt.show()
```

## 8. Basic Text Analysis - Word Frequency

```python
# Sample text analysis on a high-performing article
# Get top article
top_article_idx = df['editorial_score'].idxmax()
top_article = df.loc[top_article_idx]

print("ANALYZING TOP ARTICLE:")
print("="*80)
print(f"Path: {top_article['pagepath']}")
print(f"Editorial Score: {top_article['editorial_score']:.2f}")
print(f"Word Count: {top_article['word_count']}")
print(f"Pageviews: {top_article['pageviews']}")
print("="*80)
print(f"\nFirst 500 characters of text:")
print(top_article['full_text'][:500])
print("\n...")
```

```python
# Word frequency analysis for top article
def get_word_frequency(text, top_n=20):
    """Extract most common words from text"""
    # Convert to lowercase and split
    words = re.findall(r'\b[a-zA-Zàèéìòù]+\b', text.lower())
    
    # Remove common Italian stop words
    stop_words = {'il', 'lo', 'la', 'i', 'gli', 'le', 'un', 'uno', 'una', 'di', 'da', 'a', 'in', 
                  'su', 'per', 'con', 'tra', 'fra', 'del', 'della', 'dei', 'degli', 'delle',
                  'al', 'allo', 'alla', 'ai', 'agli', 'alle', 'dal', 'dallo', 'dalla', 'dai',
                  'e', 'o', 'che', 'è', 'sono', 'come', 'ma', 'se', 'nel', 'nella', 'nei',
                  'più', 'anche', 'quando', 'dove', 'quale', 'chi', 'cosa', 'questo', 'quello',
                  'nel', 'dello', 'negli', 'sugli', 'sulla', 'ad', 'ed', 'od'}
    
    words = [w for w in words if w not in stop_words and len(w) > 3]
    
    return Counter(words).most_common(top_n)

# Get word frequency for top article
top_words = get_word_frequency(top_article['full_text'])

print("\nMOST FREQUENT WORDS (excluding common stop words):")
print("="*60)
for word, count in top_words:
    print(f"{word:20} | {count:3} occurrences")
print("="*60)
```

```python
# Visualize word frequency
words, counts = zip(*top_words)

plt.figure(figsize=(14, 6))
plt.barh(range(len(words)), counts, color='skyblue', edgecolor='navy')
plt.yticks(range(len(words)), words)
plt.xlabel('Frequency')
plt.title(f'Top 20 Most Frequent Words in Top Article\n({top_article["pagepath"][:70]}...)', fontsize=12)
plt.gca().invert_yaxis()
plt.grid(True, alpha=0.3, axis='x')
plt.tight_layout()
plt.show()
```
